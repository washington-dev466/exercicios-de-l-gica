```python

lista = []

tamanho = int(input("Digite o tamanho da lista: "))

for i in range(tamanho):
    numero = int(input(f"Digite o número {i + 1}: "))
    lista.append(numero)

contagem = {}
for numero in lista:
    if numero in contagem:
        contagem[numero] += 1
    else:
        contagem[numero] = 1

# Organiza do mais repetido para o menos repetido
ranking = sorted(
    contagem.items(),
    key=lambda x: x[1],
    reverse=True
)

print("NÚMEROS REPETIDOS")
print("-----------------")

for numero, quantidade in ranking:
    if quantidade > 1:
        print(f"{numero} -> {quantidade} vezes")

# Descobre o maior número de repetições
maior_quantidade = ranking[0][1]

print("\nNÚMERO QUE MAIS SE REPETE")
print("--------------------------")

for numero, quantidade in ranking:
    if quantidade == maior_quantidade:
        print(f"{numero} -> {quantidade} vezes")
```
