Crie um programa que receba uma lista de números e retorne o maior valor dessa lista
1. Consultar números da lista;
2. Identificar o maior valor dentro da lista;
3. Informar números da lista e o maior dentro dela.

```python
numeros = [3, 5, 90, 2, 22]
maior = max(numeros)
print(f'Os números da lista são {numeros}.')
print(f'O maior número da lista é o {maior}.')
```

Crie um programa que receba uma lista de números e retorne a soma de todos os valores.
1. Consultar números da lista;
2. Somar os números da lista;
3. Informar o valor da soma;

```python
numeros = [1, 1, 1, 1, 1]
soma = sum(numeros)
print(f'Os números da lista são {numeros}.')
print(f'A soma dos números da lista é de {soma}.')
```

Crie um programa que receba duas listas e retorne uma terceira lista com a interseção dos elementos das duas listas.
1. Consultar a lista 1;
2. Consultar a lista 2;
3. Criar a lista 3 com os números que estão presentes nas listas 1 e 2;
4. Informar números da lista 1, 2 e 3.

```python
lista01 = [1,10,22,50,13]
lista02 = [10, 50, 12, 15, 90]
lista03 = list(set(lista01) & set(lista02))
print(f'Os números da lista 1 são {lista01}.')
print(f'Os números da lista 2 são {lista02}.')
print(f'Os números da lista 3 são {lista03}.')
```
