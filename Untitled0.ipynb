ESTRUTURA DE DADOS - DICIONÁRIOS


Criação: Crie um dicionário chamado alunos contendo 3 alunos como chaves e seus respectivos dicionários internos com nome, idade e curso.

```python
alunos = {
 "aluno1": {"nome": "Maria", "idade": 25, "curso": "Engenharia"},
 "aluno2": {"nome": "Jose", "idade": 20, "curso": "Medicina"},
 "aluno3": {"nome": "Ana", "idade": 22, "curso": "Fisica"}
}
```

Exiba no console o nome e o curso do primeiro aluno do dicionário alunos.

```python
print(alunos["aluno1"]["nome"])
print(alunos["aluno1"]["curso"])
```

Utilize o método .update() para adicionar um novo aluno ao dicionário alunos.

```python
novo_aluno = {"aluno4": {"nome": "Joao", "idade": 21, "curso": "Matematica"}}
alunos.update(novo_aluno)

print(alunos)
```
